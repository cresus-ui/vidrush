## 2. Configurer ngrok pour l'accès distant

Nous utilisons ngrok pour créer un tunnel sécurisé afin d'accéder à l'interface Streamlit depuis votre navigateur.

**Important** : Récupérez votre jeton (token) d'authentification sur le [tableau de bord ngrok](https://dashboard.ngrok.com/get-started/your-authtoken).

```python
from pyngrok import ngrok
import getpass

# Demande le token de manière sécurisée
print("Entrez votre token ngrok (collez-le ici) :")
authtoken = getpass.getpass()

ngrok.kill()
ngrok.set_auth_token(authtoken)
```

## 3. Lancer l'application et générer l'URL publique

Démarrage du serveur et création du lien d'accès :

```python
import subprocess
import time

print("🚀 Démarrage de Vidrush...")
streamlit_proc = subprocess.Popen([
    "streamlit", "run", "./webui/Main.py", "--server.port=8501"
])

time.sleep(5)

print("🌐 Création du tunnel ngrok...")
public_url = ngrok.connect(8501, bind_tls=True)

print("\n✅ Installation terminée !")
print(f"Cliquez sur ce lien pour ouvrir l'interface : {public_url}")
```
